```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar archivo JSON
with open("TelecomX_Data.json", "r") as f:
    data = json.load(f)

# 2. Convertir JSON a DataFrame
df = pd.json_normalize(data)

# 3. Renombrar columnas para simplificar
df.columns = df.columns.str.replace("customer.", "")
df.columns = df.columns.str.replace("account.", "")
df.columns = df.columns.str.replace("Charges.", "")
df.columns = df.columns.str.replace("internet.", "")
df.columns = df.columns.str.replace("phone.", "")

# 4. Convertir columnas numéricas
df["Total"] = pd.to_numeric(df["Total"], errors="coerce")
df["Monthly"] = pd.to_numeric(df["Monthly"], errors="coerce")

# 5. Información general
print("Información del dataset")
print(df.info())

print("\nValores faltantes")
print(df.isnull().sum())

print("\nEstadísticas descriptivas")
print(df.describe())

# 6. Distribución de churn
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Churn")
plt.title("Distribución de clientes que cancelaron (Churn)")
plt.show()

# 7. Churn por tipo de contrato
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Contract", hue="Churn")
plt.title("Churn según tipo de contrato")
plt.xticks(rotation=45)
plt.show()

# 8. Cargos mensuales vs churn
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Churn", y="Monthly")
plt.title("Cargos mensuales vs Churn")
plt.show()

# 9. Tenure vs churn
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Churn", y="tenure")
plt.title("Tiempo como cliente vs Churn")
plt.show()

# 10. Correlación entre variables numéricas
plt.figure(figsize=(6,4))
corr = df[["tenure","Monthly","Total"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlación entre variables numéricas")
plt.show()

# 11. Insights básicos
churn_rate = df["Churn"].value_counts(normalize=True) * 100
print("\nTasa de Churn (%)")
print(churn_rate)

avg_monthly = df.groupby("Churn")["Monthly"].mean()
print("\nPromedio de cargos mensuales por churn")
print(avg_monthly)
```
